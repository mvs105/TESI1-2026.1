# Seleção da Base de Dados do ENEM 2023

A base de dados do ENEM 2023 obtida no Kaggle passou por tratamentos prévios. Porém, para fins do projeto e aprendizado de Ciência de Dados, o objetivo é reproduzir e dominar o processo completo de preparação e limpeza de dados.

Para isso, foi executada uma **"mesclagem"** entre o conjunto de dados original e a base pré-tratada do Kaggle:
* **Recuperação de linhas:** Resgatamos as linhas da base original mantendo o critério de filtragem dos inscritos.
* **Seleção e exclusão de colunas:** Eliminamos as colunas irrelevantes ou ausentes no arquivo do Kaggle, preservando apenas os atributos de interesse para a análise.

<!-- ---

### Por que realizar este procedimento?

1. **Eficiência de Memória (RAM):** O dataset bruto original é muito extenso e pode causar estouro de memória durante o processamento[cite: 1, 3]. Filtrar registros e descartar colunas desnecessárias reduz a pegada de memória do *Dataframe*[cite: 1, 4].
2. **Qualidade e Limpeza de Dados:** Tratar nulos e inconsistências diretamente na base alinhada garante que os passos seguintes de **Estatística Descritiva** e **Visualização (EDA)** sejam precisos e confiáveis[cite: 3]. -->

```python
import pandas as pd
```

```python
df_enem_original = pd.read_csv('base_dados/MICRODADOS_ENEM_2023.csv', encoding='iso-8859-1', sep=';')
df_enem_kaggle = pd.read_csv("base_dados/MICRODADOS_ENEM_2023_trabalhado.csv")
print("Bases carregadas com sucesso!")
```

### Visualizando diferenças entre a base original e a base do Kaggle

```python
print("Quantidade de linhas e colunas na base original:")
print(df_enem_original.shape)

print("Quantidade de linhas e colunas na base do Kaggle:")
print(df_enem_kaggle.shape)
```

```python
colunas_faltantes = df_enem_original.columns.difference(df_enem_kaggle.columns)

print(list(colunas_faltantes))
print(len(colunas_faltantes))
```

### Mesclagem (Seleção de Atributos)

```python
# Se quisermos manter apenas as colunas que estão presentes na base do Kaggle e a NU_INSCRICAO, podemos fazer o seguinte:
# colunas_trabalhadas = ['NU_INSCRICAO'] + df_enem_kaggle.columns.tolist()
# df_enem_trabalhado = df_enem_original[colunas_trabalhadas]

df_enem_trabalhado = df_enem_original[df_enem_kaggle.columns]
df_enem_trabalhado.head()
```

### Seleção do DataFrame do Acre

```python
df_enem_acre = df_enem_trabalhado[df_enem_trabalhado['SG_UF_PROVA'] == 'AC']
df_enem_acre.info()
```

```python
df_enem_acre.to_csv("base_dados/MICRODADOS_ENEM_2023_AC.csv", index=False)
print("Base salva com sucesso!")
```
